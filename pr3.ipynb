```python
import matplotlib.pyplot as plt
import numpy as np

plt.subplot(2,3,1)
xpoints = np.array([1, 8])
ypoints = np.array([3, 10])
plt.plot(xpoints, ypoints, 'o')
plt.title("Title 1", color="red")
plt.xlabel("X label", color="blue")
plt.ylabel("Y label", color="blue")

plt.subplot(2,3,2)
y1 = np.array([3, 8, 1, 10])
y2 = np.array([6, 2, 7, 11])
plt.plot(y2,linestyle="dashed")
plt.plot(y1, marker = '*',color="orange")

plt.subplot(2,3,3)
ypoints = np.array([3, 8, 1, 10])
plt.plot(ypoints, marker = 'o', ms = 10, mfc = 'y',mec='b')

plt.subplot(2,3,4)
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])
plt.title("Title2")
plt.plot(x, y)
plt.grid()

plt.subplot(2,3,5)
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
size = np.full(len(x), 50)
size[1]=300
size[3]=200
size[6]=150
colors = np.array(["red","green","blue","yellow","pink","black","orange","purple","beige","brown","gray","cyan","magenta"])
plt.scatter(x, y, c=colors, s=size, alpha=0.5)


plt.subplot(2,3,6)
x=(["A","B","C","D"])
y=np.array([3,8,1,10])
plt.bar(x,y,width=0.1,color="r")

plt.show()
```
